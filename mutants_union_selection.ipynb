```python
import os
import shutil

# Function to clean QASM content by removing comments and extra spaces
def clean_qasm_content(file_path):
    cleaned_lines = []
    with open(file_path, 'r') as file:
        for line in file:
            # Remove comments and strip leading/trailing whitespace
            cleaned_line = line.split('//')[0].strip()
            if cleaned_line:  # Ignore empty lines
                cleaned_lines.append(cleaned_line)
    return '\n'.join(cleaned_lines)

# Function to compare the content of two QASM files (ignoring comments and formatting)
def are_files_equal_qasm(file1, file2):
    content1 = clean_qasm_content(file1)
    content2 = clean_qasm_content(file2)
    return content1 == content2

# Function to create unique folders with unique QASM files based on text comparison
def create_unique_files_directory_by_text(folder1, folder2, unique_folder):
    # Create the unique folder if it doesn't exist
    if not os.path.exists(unique_folder):
        os.makedirs(unique_folder)

    # Iterate through each subfolder in folder1
    for subfolder in os.listdir(folder1):
        subfolder1_path = os.path.join(folder1, subfolder)
        subfolder2_path = os.path.join(folder2, subfolder)
        unique_subfolder_path = os.path.join(unique_folder, subfolder)

        # Check if the same subfolder exists in folder2
        if os.path.isdir(subfolder1_path) and os.path.isdir(subfolder2_path):
            # Create the subfolder in the unique directory
            if not os.path.exists(unique_subfolder_path):
                os.makedirs(unique_subfolder_path)
            
            # Store the clean content of unique QASM files
            unique_files_content = []

            # Process folder1
            for file_name in os.listdir(subfolder1_path):
                src_path1 = os.path.join(subfolder1_path, file_name)
                if src_path1.endswith(".qasm") and os.path.isfile(src_path1):
                    cleaned_content = clean_qasm_content(src_path1)
                    if cleaned_content not in unique_files_content:
                        shutil.copy(src_path1, unique_subfolder_path)
                        unique_files_content.append(cleaned_content)

            # Process folder2 and check for uniqueness
            for file_name in os.listdir(subfolder2_path):
                src_path2 = os.path.join(subfolder2_path, file_name)
                if src_path2.endswith(".qasm") and os.path.isfile(src_path2):
                    cleaned_content = clean_qasm_content(src_path2)
                    if cleaned_content not in unique_files_content:
                        shutil.copy(src_path2, unique_subfolder_path)
                        unique_files_content.append(cleaned_content)

    print(f"Unique QASM files copied to {unique_folder}")

# Usage
folder1 = r'all_mutants\muskit'
folder2 = r'all_mutants\qmutpy'
unique_folder = r'all_mutants\unique_mutants'

create_unique_files_directory_by_text(folder1, folder2, unique_folder)
```

```python
import os
import shutil

def group_files_by_name(base_folder):
    # Define the grouping rules
    keyword_mapping = {
        'Add': ['Add', 'QGI'],
        'Remove': ['Remove'],
        'Replace': ['Replace', 'QGR'],
        'QMD': ['QMD'],
        'QMI': ['QMI']
    }
    
    # Loop through each subfolder in the base folder
    for subfolder in os.listdir(base_folder):
        subfolder_path = os.path.join(base_folder, subfolder)
        
        if os.path.isdir(subfolder_path):  # Only process directories
            # Create sub-subfolders for each keyword group
            for keyword in keyword_mapping.keys():
                keyword_folder = os.path.join(subfolder_path, keyword)
                os.makedirs(keyword_folder, exist_ok=True)
            
            # Loop through files in the subfolder
            for filename in os.listdir(subfolder_path):
                file_path = os.path.join(subfolder_path, filename)
                
                if os.path.isfile(file_path):  # Only process files
                    # Check which group the filename belongs to based on the mapping
                    for folder_name, keywords in keyword_mapping.items():
                        if any(keyword in filename for keyword in keywords):
                            # Move the file to the corresponding sub-subfolder
                            target_folder = os.path.join(subfolder_path, folder_name)
                            shutil.move(file_path, target_folder)
                            break  # File has been moved, no need to check further

# Example usage:
group_files_by_name(r"all_mutants\unique_mutants")
```

```python
import os
import shutil
import random

def copy_random_files(base_folder, destination_folder, num_files):
    # Loop through each subfolder in the base folder
    for subfolder in os.listdir(base_folder):
        subfolder_path = os.path.join(base_folder, subfolder)
        
        if os.path.isdir(subfolder_path):  # Only process directories
            # Create the corresponding subfolder in the destination folder
            dest_subfolder_path = os.path.join(destination_folder, subfolder)
            os.makedirs(dest_subfolder_path, exist_ok=True)
            
            # Loop through sub-subfolders in the current subfolder
            for subsubfolder in os.listdir(subfolder_path):
                subsubfolder_path = os.path.join(subfolder_path, subsubfolder)
                
                if os.path.isdir(subsubfolder_path):  # Only process sub-subfolders
                    # Get a list of all files in the sub-subfolder
                    files = [f for f in os.listdir(subsubfolder_path) if os.path.isfile(os.path.join(subsubfolder_path, f))]
                    
                    # Randomly select num_files files (or fewer if not enough files)
                    selected_files = random.sample(files, min(len(files), num_files))
                    
                    # Copy each selected file to the destination subfolder (no sub-subfolder creation)
                    for file in selected_files:
                        src_file_path = os.path.join(subsubfolder_path, file)
                        dest_file_path = os.path.join(dest_subfolder_path, file)
                        shutil.move(src_file_path, dest_file_path)

# Example usage:
copy_random_files(r"all_mutants\unique_mutants", r"all_mutants\selected_mutants", 5)
```
