```python
import os
import pandas as pd

def find_csv_files(directory):
    # List to hold the rows of the new DataFrame
    data = []
    
    # Walk through the directory
    for root, _, files in os.walk(directory):
        for file in files:
            # Check if the file is a .qasm file
            if file.endswith('.csv') and 'solutions' in file:
                # Construct the full path to the .qasm file
                solutions_csv = os.path.join(root, file)
                origin_qc = root.split('mutants_')[-1]
                
                gates_csv = solutions_csv.replace(f'best_solutions', f'suspicious_gates')
                
                if 'Add' in file or 'Remove' in file or 'Replace' in file:
                    splited = file.split('_')
                    mutation_pos = int(splited[8].replace('P',''))
                    mutant_operator = splited[6]
                    qubits = int(splited[4])
                else:
                    splited = file.split('_')
                    mutation_pos = int(splited[-1].replace('P','').replace('.csv',''))
                    mutant_operator = splited[6]
                    qubits = int(splited[5])
                    
                if qubits == 2:
                    total_positions = 18
                elif qubits == 3:
                    total_positions = 18
                elif qubits == 4:
                    total_positions = 49
                elif qubits == 5:
                    total_positions = 55
                elif qubits == 6:
                    total_positions = 37
                elif qubits == 7:
                    total_positions = 41
                elif qubits == 8:
                    total_positions = 55
                elif qubits == 9:
                    total_positions = 59
                elif qubits == 10:
                    total_positions = 48
                elif qubits == 11:
                    total_positions = 43
                elif qubits == 12:
                    total_positions = 37
                elif qubits == 13:
                    total_positions = 27
                    
                #print(solutions_csv)
                # Check if the corresponding .csv file exists
                if os.path.exists(solutions_csv) and os.path.exists(gates_csv):
                    # Read the first row of the CSV file
                    first_row_solutions = pd.read_csv(solutions_csv, nrows=1)
                    gates_df = pd.read_csv(gates_csv)
                    if mutation_pos in gates_df['Position'].values:
                        sus_pos = (gates_df[gates_df['Position']==mutation_pos].index[0] / total_positions) * 100
                    else:
                        sus_pos = 100
                    row_data = {'origin_qc': origin_qc, 'filename': file,'mutation_pos': mutation_pos, 'mutation_operation':mutant_operator, 'Score': first_row_solutions['Score'].iloc[0], 'Fix_operation': first_row_solutions['Operator'].iloc[0], 'Fixed_gate':first_row_solutions['Gate'].iloc[0], 'New_gate':first_row_solutions['New_gate'].iloc[0], 'Fix_pos':first_row_solutions['Position'].iloc[0], 'Mutation_sus_pos': sus_pos, 'Exec_time':  first_row_solutions['Total_exec_time'].iloc[0], 'Overall_improvement_percent':first_row_solutions['Overall_improvement_percent'].iloc[0]}
                    data.append(row_data)
    
    # Create a new DataFrame from the collected data
    df = pd.DataFrame(data)
    
    # Manually add the results obtained from the Bugs4Q benchmark
    df.loc[len(df)] = ["bugs4q-id25", "best_solutions_for_Bug4Q_id25.csv", 3, "SQ", 7.637732, "Replace", "ccx", "r", 2, 50, 431.749625, 3.589870]
    df.loc[len(df)] = ["bugs4q-id26", "best_solutions_for_Bug4Q_id26.csv", 2, "Replace", 0, "Replace", "h", "rx", 1, 0, 17.291166, 100]
    df.loc[len(df)] = ["bugs4q-id27", "best_solutions_for_Bug4Q_id27.csv", 8, "Reverse", 0, "Add", "t", "swap", 7, 76.470588, 82.627925, 100]
    df.loc[len(df)] = ["bugs4q-id39", "best_solutions_for_Bug4Q_id39.csv", 0, "Delete", 23.646314, "Add", "cx", "h", 0, 90, 2437.438162, 10.620157]
    
    df.to_csv('results_summary.csv')
    
    return df

```
